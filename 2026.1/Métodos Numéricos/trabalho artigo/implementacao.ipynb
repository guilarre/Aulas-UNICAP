```python
import sympy as sp

def ponm_solver(func_str, x0, tol=1e-15, max_iter=100):
    """
    Implementação do Método Ótimo de Quarta Ordem (PONM)
    Baseado em Qureshi et al. (2024), fórmula (18).
    
    Parâmetros:
    func_str (str): A função psi(x) como string.
    x0 (float): Estimativa inicial.
    tol (float): Tolerância para convergência.
    max_iter (int): Número máximo de iterações.
    """
    x = sp.symbols('x')
    psi_expr = sp.sympify(func_str)
    psi_prime_expr = sp.diff(psi_expr, x)
    
    # Conversão para funções lambda para avaliação rápida
    psi = sp.lambdify(x, psi_expr, 'numpy')
    psi_prime = sp.lambdify(x, psi_prime_expr, 'numpy')
    
    xn = x0
    print(f"{'n':<5} | {'xn':<20} | {'|psi(xn)|':<20}")
    print("-" * 50)
    
    for n in range(max_iter):
        val_psi_xn = psi(xn)
        val_psi_prime_xn = psi_prime(xn)
        
        if abs(val_psi_prime_xn) < 1e-20:
            print("Erro: Derivada nula ou muito próxima de zero.")
            break
            
        # Passo intermediário (Newton): yn = xn - psi(xn)/psi'(xn)
        yn = xn - val_psi_xn / val_psi_prime_xn
        val_psi_yn = psi(yn)
        
        # Fórmula de quarta ordem ótima (Formula 18)
        # xn+1 = xn - [psi(xn)*(psi(xn) - 2*psi(yn)) - psi(yn)**2] / 
        #             [psi'(xn)*(psi(xn) - 3*psi(yn))]
        
        num = val_psi_xn * (val_psi_xn - 2 * val_psi_yn) - (val_psi_yn**2)
        den = val_psi_prime_xn * (val_psi_xn - 3 * val_psi_yn)
        
        xn_plus_1 = xn - (num / den)
        
        print(f"{n:<5} | {xn:<20.15f} | {abs(val_psi_xn):<20.5e}")
        
        # Critério de convergência
        if abs(xn_plus_1 - xn) < tol or abs(psi(xn_plus_1)) < tol:
            return xn_plus_1
            
        xn = xn_plus_1
        
    return xn

# Exemplo de uso: Equação de Van der Waals (Problema 1 do artigo)
vdw_eq = "40*x**3 - 95.26535116*x**2 + 35.28*x - 5.6998368"
raiz_vdw = ponm_solver(vdw_eq, x0=1.5)
print(f"\nRaiz encontrada: {raiz_vdw}")
```
