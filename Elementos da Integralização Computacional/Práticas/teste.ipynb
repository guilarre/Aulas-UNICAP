```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
# from IPython.display import display, Math

sp.init_printing(order='none')
```

```python
x = sp.Symbol('x')
f1 = sp.exp(x)
f2 = x+2

expression = sp.Eq(f1, f2)
print(str(sp.nsolve(expression, -1.5)) + ", " + str(sp.nsolve(expression, 1.5)))
```
